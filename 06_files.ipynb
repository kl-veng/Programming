# Working with files

Before a file can be read from or written to, it must be opened. A modern way to do this is using a context manager.

## Open a file in a context manager
It is recommended to use context manager, which not only opens for us the file, but also closes it. The syntax is as follows:

````python
with open('data/names/names_short.txt', encoding='utf-8') as fh:
    # Do something with the file 
    # (as we did not say what to do with the file, we will get an error now)
````

The object representing the open file provides several ways to access the contents of the file, including an iterator that we can use in a `for` loop. 


````python
with open('data/names/names_short.txt', encoding='utf-8') as fh:
    for line in fh:
        print(line)
````

In some older scripts, you might still see the following opening and closing of the file, however, this is not recommended.

````python
fh = open('data/names/names_short.txt', encoding='utf-8')
# Do something with the file
fh.close()
````

## Absolute and Relative Paths

A **path** tells Python where to find a file or folder.

**Relative path:**
A relative path describes the location **relative to the current working directory**.

```python
'data/names/names_short.txt'
```

This means: go into the `data` folder, then into `names`, and find `names_short.txt`.

**Absolute path:**
An absolute path gives the **complete location**, starting from the root of the file system.

For example, on Windows:

```python
"C:/Users/klara/Desktop/Programming/data/names/names_short.txt"
```

The important difference is:

> **A relative path depends on where your program is currently running; an absolute path always points to the same location. Relative paths are shorter and more portable; absolute paths are unambiguous but usually specific to one computer.**


### A note about Windows paths

Windows traditionally uses **backslashes (`\`)** in paths:

```text
C:\Users\klara\Desktop\Programming\data\names\names_short.txt
```

However, in Python, backslashes have a special meaning because they introduce **escape sequences**. For example, `\n` means a newline.

Therefore, using forward slashes is often easier:

```python
"C:/Users/klara/Desktop/Programming/data/names/names_short.txt"
```

Python accepts double forward slashes in Windows paths, too:

```python
"C:\\Users\\klara\\Desktop\\Programming\\data\\names\\names_short.txt"
```

Alternatively, you can use a **raw string**:

```python
r'C:\Users\klara\Desktop\Programming\data\names\names_short.txt'
```

The `r` tells Python to treat the backslashes as ordinary characters rather than escape sequences.

## Other methods to read from a file

### read()
The `read()` method reads the entire file contents as a string. So we get the whole file content as a (sometimes very long) string:

````python
with open('data/names/names_short.txt', encoding='utf-8') as fh:
    data = fh.read()
print(data)    
````

### readlines()
This method reads each line of the file as an element in a list (a list is another sequence type we will learn about soon):

````python
with open('data/names/names_short.txt', encoding='utf-8') as fh:
    lines = fh.readlines()
print(lines)    
````

## Write to a file
So far we have only read from files. To be able to write to a file, we need to open it in a special way. The `open()` function expects as second argument a string specifying how to open a file. If we do not specify, the default value `r` (for `read`) is assumed.

```
with open('data/first_name/names_short.txt', encoding='utf-8') as fh:
```
leads to the same result as 

```
with open('data/first_name/names_short.txt', 'r', encoding='utf-8') as fh:
```


If we want to open a file for writing, we use `'w'` (for `write`) instead of `'r'`.
If we want to append new text to an existing file, we use  `'a'` (for  `append`)

````python
with open('data/testfile.txt', 'a', encoding='utf-8') as fh:
    fh.write('I am a text.')
````

## Write binary data
So far we have always assumed that we are reading text from a file or writing to a file. If we're dealing with binary data (i.e. anything that isn't plain text, e.g. images, PDF files, Word files, etc.), we need to specify that explicitly with the letter `'b'`:

````python
with open('img/string1.png', 'rb') as fh_in:
    data = fh_in.read()

with open('img/testimage.png', 'wb') as fh_out:
    fh_out.write(data)
````
