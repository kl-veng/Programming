# How to work with Jupyter Notebooks

## What is a Jupyter notebook?

A notebook is an interactive Python shell running in the browser. Not only can Python snippets be entered and executed immediately in a notebook, it is also possible to enrich the program code with text, images, videos, and so on. Notebooks are therefore a great way to create a kind of interactive script.

Jupyter Notebooks are saved with all content into a single file (extension: .ipynb) and can therefore be easily shared. Furthermore, there is the possibility to export the notebooks into different formats (in the browser under File -> Download as, in VS Code via the "Export as" icon ).

## Jupyter Notebooks and Python scripts

Jupyter Notebooks are beneficial wherever you need to integrate program code with text and run small program snippets directly from this environment -- for example, as an interactive script for that course or in science, where you can interactively edit the data under investigation, visualize it and document the results right away. Their interactive nature also makes them a great tool to learn programming.

However, this should not stop you from also practicing the traditional and much more widespread form of programming:

- Source code is written in an editor or Integrated Development Environment (IDE) and saved as a file (in Python, typically with a .py filename extension).

- This file is passed to a Python interpreter for execution.

## How to run a Jupyter Notebook

Prerequisite for this step is that you have installed Python and Jupyter on your computer (if you followed the installation with Anaconda, both is installed automatically).

Enter the following command to start the notebook server in the terminal:

jupyter lab

This command starts the server and opens the browser where the dashboard is displayed. If this does not happen automatically, in the output of the starting notebook server you will find the URL where you can reach it in the browser (starts with http://localhost:...).

## The dashboard

You can open an existing document or create a new one by clicking the + button. In addition, there are a number of other options, which we will not go into here (yet).



## Working with a Notebook

If you have opened an existing document or created a new one, you will see a menu bar similar to a normal program, below it the toolbar and below that the actual workspace organized in cells.



## The Jupyter workspace: cells



A notebook is organized into individual cells. Each cell is either

- **Code**  , i.e. program code
- **Markdown**, i.e. text possibly with multimedia elements
- **Raw NB Convert** defines a "Raw Cell", where you can define content e.g. only for certain output formats.

The first two options are relevant for us.




## Code cells

In a code cell we can directly enter and execute Python code. Set the focus on the next cell (the cell with the content "Hello world"), e.g. by clicking into it with the mouse and execute the content. You have several possibilities to execute the code:

- Click on `▶` symbol in the toolbar
- Click on `Run` in the menu and then `Run Selected Cell`
- Press the key combination `Ctrl + Enter`, i.e. hold down the `Ctrl` key and press the `Enter` key at the same time.
- Press the key combination `Alt + Enter`, i.e. hold down the `Alt` key and press the `Enter` key at the same time. This not only executes the code, but also immediately creates a new blank cell below the executed cell.

## Markdown Cells

Markdown cells contain connecting or explanatory texts. You can also change these cells at any time or, for example, extend them with your own notes. To do this, simply double-click in the cell. This will make it editable. To return to normal mode, execute the cell as described above for code cells.

Markdown is a very quick to learn markup language that you can use to structure texts. You can find a good introduction e.g. here: https://www.earthdatascience.org/courses/intro-to-earth-data-science/file-formats/use-text-files/format-text-with-markdown-jupyter-notebook/.

Markdown is not only the language of Jupyter Notebooks, but is also preferred for documentation of software projects but also for many other purposes.

## Insert new cells

You can insert new cells anytime and anywhere, e.g. for your own notes or your own code experiments. To do this, click on the + symbol in the toolbar or use the Edit --> Paste Cell menu item. Using the keyboard, you can press A or B to insert a new cell above or below the current cell.
